```python
import yfinance as yf
stocks = ["INFY.NS", "TCS.NS", "RELIANCE.NS"]

for s in stocks:
    data = yf.download(s, period="1y")
    data.columns=data.columns.get_level_values(0)
    data=data.reset_index()
    data.to_csv(f"{s}.csv", index=False)
```

```python
data = data.dropna() 
```

```python
data.info()
```

```python
data['Daily Return'] = (data['Close'] - data['Open']) / data['Open']
```

```python
data['MA_7'] = data['Close'].rolling(window=7).mean()
```

```python
high_52 = data['High'].max()
low_52 = data['Low'].min()
```

```python
data = data.sort_values(by='Date')
```

```python
data['Volatility'] = data['Daily Return'].rolling(window=7).std()
```
